# Activity 4.02: Random Forest Classification for Your Car Rental Company

## Introduction

The goal of this activity is to build and evaluate Random Forest and Extra Trees classification models using the Car Evaluation dataset. These models predict whether a car is unacceptable, acceptable, good, or very good.

The models are evaluated using accuracy, confusion matrices, classification reports, and feature importance scores. The results help identify the most important factors when selecting cars for the company's fleet.

```python

# Step 1: Import the necessary libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# Step 2: Load the Car Evaluation dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"

columns = [
    "buying", "maint", "doors", "persons",
    "lug_boot", "safety", "class"
]

df = pd.read_csv(url, header=None, names=columns)

# Display the first five rows
display(df.head())

# Check the dataset dimensions and data types
print("Dataset shape:", df.shape)
print(df.info())

# Check for missing values
print(df.isnull().sum())
```

```python

# Step 3: Encode the categorical values into numerical values
encoder = LabelEncoder()

for column in df.columns:
    df[column] = encoder.fit_transform(df[column])

# Extract the target variable using pop()
labels = df.pop("class")

# Separate features and target into training and testing sets
features_train, features_test, label_train, label_test = train_test_split(
    df,
    labels,
    test_size=0.10,
    random_state=42,
    stratify=labels
)

print("Training features:", features_train.shape)
print("Testing features:", features_test.shape)
print("Training labels:", label_train.shape)
print("Testing labels:", label_test.shape)
```

```python

# Step 4: Create and train the Random Forest classifier
random_forest_classifier = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
random_forest_classifier.fit(features_train, label_train)

# Predict the test labels
rf_predictions = random_forest_classifier.predict(features_test)

# Evaluate the model
print("Random Forest Accuracy:",
      accuracy_score(label_test, rf_predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(label_test, rf_predictions))

print("\nClassification Report:")
print(classification_report(
    label_test,
    rf_predictions,
    zero_division=0
))

# Display feature importance
print("\nRandom Forest Feature Importance:")
print(random_forest_classifier.feature_importances_)

print("\nFeature Names:")
print(features_train.columns.tolist())
```

```python

# Step 5: Create and train the Extra Trees classifier
extra_trees_classifier = ExtraTreesClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
extra_trees_classifier.fit(features_train, label_train)

# Predict the test labels
et_predictions = extra_trees_classifier.predict(features_test)

# Evaluate the model
print("Extra Trees Accuracy:",
      accuracy_score(label_test, et_predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(label_test, et_predictions))

print("\nClassification Report:")
print(classification_report(
    label_test,
    et_predictions,
    zero_division=0
))

# Display feature importance
print("\nExtra Trees Feature Importance:")
print(extra_trees_classifier.feature_importances_)

print("\nFeature Names:")
print(features_train.columns.tolist())
```

```python

# Step 6: Compare both models
comparison = pd.DataFrame({
    "Model": ["Random Forest", "Extra Trees"],
    "Accuracy": [
        accuracy_score(label_test, rf_predictions),
        accuracy_score(label_test, et_predictions)
    ]
})

display(comparison)
```

### Model Comparison and Results

The Random Forest classifier achieved an accuracy of 98.27%, while the Extra Trees classifier achieved an accuracy of 98.84% on the test dataset. Extra Trees performed slightly better, with an improvement of approximately 0.58 percentage points.

Both models achieved high accuracy, suggesting that they learned the patterns in the Car Evaluation dataset effectively. However, accuracy alone does not provide a complete evaluation, so the confusion matrices and classification reports should also be considered when comparing model performance.

### Conclusion

The Random Forest classifier achieved an accuracy of 98.27%, while the Extra Trees classifier achieved an accuracy of 98.84%. Extra Trees performed slightly better on the test dataset.

The confusion matrices and classification reports show that both models correctly classified most of the cars. The feature importance results indicate that safety and persons were among the most important features for both models.

Overall, both models performed well, but Extra Trees achieved the highest accuracy in this experiment. Further testing would help determine whether this performance generalizes to other datasets.
