# Activity 4.01: Car Data Classification

## Introduction

The goal of this activity is to build and evaluate a Decision Tree classifier using the Car Evaluation dataset. The model classifies cars into four categories: unacceptable, acceptable, good, and very good. The model will be evaluated using its accuracy score and classification report.

```python
# Step 1: Import the necessary libraries

import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report

# Load the Car Evaluation dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"

columns = [
    "buying", "maint", "doors", "persons",
    "lug_boot", "safety", "class"
]

df = pd.read_csv(url, header=None, names=columns)

# Display the first five rows
display(df.head())

# Check the dataset
print("Dataset shape:", df.shape)
df.info()

# Check for missing values
print(df.isnull().sum())
```

```python
# Step 2: Perform label encoding

encoder = LabelEncoder()

for column in df.columns:
    df[column] = encoder.fit_transform(df[column])

# Display the encoded data
display(df.head())
```

```python
# Step 3: Extract the label variable using pop()

labels = df.pop("class")

# Step 4: Split the data into training and testing sets

features_train, features_test, label_train, label_test = train_test_split(
    df,
    labels,
    test_size=0.10,
    random_state=42,
    stratify=labels
)

print("Training features:", features_train.shape)
print("Testing features:", features_test.shape)
print("Training labels:", label_train.shape)
print("Testing labels:", label_test.shape)
```

```python
# Step 5: Build and train the Decision Tree classifier

decision_tree_classifier = DecisionTreeClassifier(random_state=42)

# Train the model
decision_tree_classifier.fit(features_train, label_train)

# Make predictions
labels_predicted = decision_tree_classifier.predict(features_test)

print("The Decision Tree model has been trained successfully.")
```

```python
# Step 6: Check the model score

accuracy = decision_tree_classifier.score(features_test, label_test)

print("Decision Tree Accuracy:", accuracy)
print("Decision Tree Accuracy: {:.2%}".format(accuracy))
```

```python
# Step 7: Evaluate the model using a classification report

print("Classification Report:")

print(
    classification_report(
        label_test,
        labels_predicted,
        labels=[0, 1, 2, 3],
        target_names=["acc", "good", "unacc", "vgood"],
        zero_division=0
    )
)
```

## Results and Conclusion

The Decision Tree classifier was trained using 90% of the Car Evaluation dataset and tested using the remaining 10%. The model achieved an accuracy of 99.42%.

The classification report showed strong precision, recall, and F1-score across the four car categories. However, the recall for the very good category was 0.86, meaning that the model did not correctly identify every car in this category.

In conclusion, the Decision Tree classifier performed very well on the test data. It classified cars based on buying price, maintenance cost, number of doors, passenger capacity, luggage space, and safety level.
